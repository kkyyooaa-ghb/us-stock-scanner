## tl;dr

- 2026-07-27 的 138 筆 Notion 樣本中，舊引擎已定案 20 筆，V1.2.0 僅 2 筆。
- 合併後的 22 筆通過總量門檻，但 V1.2.0 進度只有 2/15，不能啟動權重或門檻調整。
- 正確下一步是修正週報分代閘門並繼續收集 V1.2.0 R 樣本。

## Context & Methods

決策問題：7/26 週報是否已提供足夠證據校準 V1.2.0 計分核心。

### Key Assumptions

- `Status` 以 `📐`、`📉`、`🌤️` 開頭者屬 V1.2.0；以 `🔥` 開頭者屬舊引擎。
- D8 的 15 筆門檻必須在同一策略世代內計算。
- CSV 是 2026-07-27 由 Notion 資料庫即時 SQL 查詢保存的聚合快照。

## Data

```python
from csv import DictReader
from pathlib import Path

source_path = Path('.scratch/v12-calibration-gate/notion-cohort-summary-2026-07-27.csv')
with source_path.open(encoding='utf-8', newline='') as source_file:
    cohorts = list(DictReader(source_file))

for row in cohorts:
    for field in ('n_total', 'n_r', 'wins', 'stops'):
        row[field] = int(row[field])
    row['r_mean'] = float(row['r_mean'])

cohorts
```

## Results

```python
calibration_min_r = 15
by_engine = {row['engine']: row for row in cohorts}
v12 = by_engine['V1.2.0']
legacy = by_engine['legacy']

result = {
    'all_finalized': v12['n_r'] + legacy['n_r'],
    'all_samples': v12['n_total'] + legacy['n_total'],
    'v12_progress': f"{v12['n_r']}/{calibration_min_r}",
    'v12_ready': v12['n_r'] >= calibration_min_r,
    'legacy_finalized': legacy['n_r'],
}
result
```

```python
assert result['all_finalized'] == 22
assert result['all_samples'] == 138
assert result['v12_progress'] == '2/15'
assert result['v12_ready'] is False
result
```

## Takeaways

總樣本門檻與新引擎校準門檻不能互換。20 筆舊引擎 R 值只適合作為 baseline；目前 V1.2.0 的兩筆皆為 -1R，樣本量與結果都只支持繼續觀察。
